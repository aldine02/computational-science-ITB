```python
import numpy as np
from scipy.integrate import solve_ivp
import pandas as pd
from matplotlib import pyplot as plt
```

```python
e = np.e
t = np.arange(0, 10, 0.1)
```

Persamaan A. $$\frac{dP}{dt} = 0.01P$$

```python
def P(t, P):
  return 0.01*P

sol = solve_ivp(P, (0, 10), [10], method='RK45', t_eval=t)
```

```python
df = pd.DataFrame({"Waktu": sol.t, "Populasi": sol.y[0]})
df.head(11)
```

```python
plt.plot(sol.t, sol.y[0])
plt.xlabel("Waktu")
plt.ylabel("Populasi")
plt.show()
```

# Soal 2

## Carrying Capacity ($L$)

Total beras tersedia:

$$110 \frac{bu}{year} \times 45 lb \times 100 acres $$

$$= 495000 \frac{lb}{year}$$

Kebutuhan 1 mallard per tahun:
$$\frac{3.2\frac{oz}{day}}{16oz} = 0.2 \frac{lb}{day}$$

$$0.2 \times 365 = 73 \frac{lb}{year}$$

Carrying Capacity ($L$):

$$L = \frac{\text{Total beras tersedia}}{\text{Kebutuhan 1 Mallard per tahun}}$$

$$L = \frac{495000 \frac{lb}{year}}{73 \frac{lb}{year}}$$

$$= 6780.82$$



Persamaan pertumbuhan terbatas:

$$\frac{dP}{dt} = kP \left (1 - \frac{P}{L} \right)$$

dimana:

*Carrying Capacity* ($L$) = 6780.82

## Program (Mallard):
Asumsi:

$k$ = 2

Populasi Awal = 10

```python
k = 2
L = 6780.82
populasi_awal = 10

def P(t, P):
  return k*P*(1 - P/L)

sol = solve_ivp(P, (0, 10), [populasi_awal], method='RK45', t_eval=t)

plt.plot(sol.t, sol.y[0])
plt.xlabel("Waktu")
plt.ylabel("Populasi Mallard")
plt.show()
```

Populasi Mallard meningkat dari 10 menuju limit *carrying capacity*
