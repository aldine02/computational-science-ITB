Approximate:
$$\int_0^2 3x^2 dx$$

```python
def f(x):
  return 3*x**2

class Integration():
  def __init__(self, f, a, b):
    self.f = f
    self.a = a
    self.b = b
```

Analytic Answer:

$$\int_0^2 3x^2 dx = |_0^2 x^3 = 2^3 = 8$$

```python
ref = 8
```

Right-Riemann Sum:

$$\int_a^b f(x) dx \approx \Delta x [f(x_0) + f(x_1) + ... + f(x_{n-1})]$$

```python
import numpy as np

class RRS(Integration):
  def __call__(self, n):
    s = 0
    dx = (self.b-self.a)/n
    x = np.linspace(self.a,self.b,n)
    for i in range(0, len(x)):
      s += self.f(x[i])
    s *= dx
    return s
```

Trapezoidal Rule:

$$\int_a^b f(x) dx \approx \frac{\Delta x}{2} [f(x_0) + 2f(x_1) + ... + 2f(x_{n-1}) + f(x_n)]$$

```python
class Trapezoidal(Integration):
  def __call__(self, n):
    s = 0
    dx = (self.b-self.a)/n
    x = np.linspace(self.a,self.b,n)
    s += f(x[0]) + f(x[-1])
    for i in range(1, len(x)-1):
      s += f(x[i]) + f(x[i+1])
    s *= dx/2
    return s
```

Simpson's Rule:

$$\int_a^b f(x) dx \approx \frac{\Delta x}{3} [f(x_0) + 4f(x_1) + 2f(x_2) + 4f(x_3) + 2f(x_4) + ... + 4f(x_{n-1}) + f(x_n)] $$

```python
class Simpson(Integration):
  def __call__(self, n):
    s = 0
    dx = (self.b-self.a)/n
    x = np.linspace(self.a,self.b,n)
    s += f(x[0]) + f(x[-1])
    for i in range(1, len(x)-1):
      if i % 2 == 0:
        s += 2 * f(x[i])
      else:
        s += 4 * f(x[i])
    s *= dx/3
    return s
```

Make a table containing the approximated values and errors of each method for $n = 10, 10^2, 10^3, 10^4$

```python
n_points = [10, 100, 1000, 10000]

print('   n              Right-Reimann Sum                            Trapezoidal Rule                           Simpson\'s Rule')

for n in n_points:
  rrs = RRS(f,0,2)(n)
  e1 = np.abs(rrs - ref)

  trap = Trapezoidal(f,0,2)(n)
  e2 = np.abs(trap - ref)

  simp = Simpson(f,0,2)(n)
  e3 = np.abs(simp - ref)

  print(f'{n:5d}     approx={rrs:.8f}, error={e1:.8f}        approx={trap:.8f}, error={e2:.8f}       approx={simp:.8f}, error={e3:.8f}')
```
